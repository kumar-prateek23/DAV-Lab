```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA

# Load the dataset
df = pd.read_csv(r"C:\Users\User\Downloads\archive (11)\winequality-red.csv")

# Display the first five rows
print("First five rows:")
print(df.head())
```

```python
print(df.shape)
```

```python
print("Dataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nDataset information:")
print(df.info())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
x=df.drop("quality", axis=1)
y=df["quality"]
```

```python
# Separate input features and target variable

X = df.drop("quality", axis=1)
y = df["quality"]

# Display first five rows
print("Input features (X):")
print(X.head())

print("\nTarget variable (y):")
print(y.head())
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_normalized = pd.DataFrame(
    scaler.fit_transform(X),
    columns=X.columns
)

print("Normalized input features:")
print(X_normalized.head())
```

```python
from sklearn.preprocessing import StandardScaler

# Create StandardScaler object
scaler = StandardScaler()

# Standardize input features
X_standardized = pd.DataFrame(
    scaler.fit_transform(X),
    columns=X.columns
)

# Display first five rows
print("Standardized dataset:")
print(X_standardized.head())
```

```python
from sklearn.decomposition import PCA

# Create PCA object with 2 components
pca = PCA(n_components=2)

# Apply PCA for dimensionality reduction
X_pca = pca.fit_transform(X_standardized)

# Convert PCA result into DataFrame
df_pca = pd.DataFrame(
    X_pca,
    columns=["PC1", "PC2"]
)

# Display first five rows
print("PCA transformed dataset:")
print(df_pca.head())

# Display new dataset shape
print("\nShape after PCA:")
print(df_pca.shape)
```

```python
# Print explained variance ratio
print(pca.explained_variance_ratio_.round(4))
```

```python
print("Total variance retained:")
print(pca.explained_variance_ratio_.sum().round(4))
```

```python
import matplotlib.pyplot as plt

# Visualize PCA results
plt.figure(figsize=(8, 5))

plt.scatter(
    df_pca["PC1"],
    df_pca["PC2"],
    c=y,
    cmap="viridis",
    alpha=0.7
)

plt.title("PCA Visualization - Wine Quality Dataset")
plt.xlabel("Principal Component 1 (PC1)")
plt.ylabel("Principal Component 2 (PC2)")

plt.colorbar(label="Wine Quality")

plt.grid(True)
plt.show()
```
