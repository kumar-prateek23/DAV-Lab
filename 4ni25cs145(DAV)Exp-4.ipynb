```python
df_mean=df.copy()
```

```python
mean_building_area=df_mean["buildingArea"].mean()
```

```python
df_mean["buildingArea"] = df_mean["buildingArea"].fillna(mean_building_area)

print("Missing values in buildingArea after filling:")
print(df_mean["buildingArea"].isnull().sum())
```

```python
mean_building_area = df_mean["BuildingArea"].mean()

df_mean["BuildingArea"] = df_mean["BuildingArea"].fillna(mean_building_area)

print("Missing values after filling:")
print(df_mean["BuildingArea"].isnull().sum())
```

```python
df_median = df.copy()

median_building_area = df_median["BuildingArea"].median()

df_median["BuildingArea"] = df_median["BuildingArea"].fillna(median_building_area)

print("Missing values after median filling:")
print(df_median["BuildingArea"].isnull().sum())
```

```python
df_ffill=df.copy()
```

```python
df_ffill = df_ffill.ffill()

print("Missing values after forward fill:")
print(df_ffill.isnull().sum())
```

```python
df_bfill = df.copy()

df_bfill = df_bfill.bfill()

print("Missing values after backward fill:")
print(df_bfill.isnull().sum())
```
